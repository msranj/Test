#### Conditional statements  

1. if .. else loop
2. While loop
3. For Loop



```python
# If.. else 
# case: if its raining, then take an umbrella, else take shorts

# if.. else statement
if (1 < 1):
    print("True statement")
else:
    print("Not a True statement")


# note: If and else statements should be at same line/level of indentation. If not, it will throw an error.
# there are other relational operators like 
    # ==, 
    # !=, 
    # >=, 
    # <=, 
    # >, 
    # < 
# which can be used in if statements.
# note: " = " in python is used for assignment of a value.

```

```python
# elif statement

myage = 13

if (myage <= 13):
    print("You are a child")

elif (myage > 13 and myage <=18):
    print("You are a teenager")

elif (myage > 18 and myage <= 35):
    print("You are an young adult")
    
elif (myage > 35 and myage <= 50):
    print("You are a middle-aged adult")

else:
    print("You are an Senior Citizen")
```

```python
# For loops

# basically we are iterating on a list of items in a python data structure like tuples, lists.

list_fruits = ["apple","banana", "mango", "grapes"]
tuple_rates_in_rupees = (100,200,300,150)

print("The list of fruits are:")
for item in list_fruits:
    
    print(item)

print("The price of each fruits are:")
for amount in tuple_rates_in_rupees:
    print(amount)


# lets see Range function in for loops

for i in range(0,10):
    print(i)    

# lets see skip value in range function i.e incremental factor
# syntax: range(start, stop, incremental factor)

print("The even numbers between 0 and 10 are:")
for i in range (0,11,5):
    print(i)


```

```python

# print first 10 multiples of 5 using range.
print("The first 10 multiples of 5 are:")
for i in range(5,50,5):
    print(i)

```

```python
# try to print tables of 5
for i2 in range(0,51,5):
    print(i2)
```

```python
# nestes for loops
for i in range(0,10,2):
    for j in range(0,3):
        print(i*j)
        
```

```python
# while loops:
# these run as long as condition is true, once enters false status, it terminates
# used when we don't know the number of iterations in advance.

a = 0

while a < 10:
    print(a)
    a = a+5
    # we increment or decrement the condition, else the above condition is always true and run endlessly.

# there are 3 control statements in python.
    # 1. break: it breaks the loop and comes out of the loop.
    # 2. continue: it skips the current iteration and goes to next iteration.
    # 3. pass: it is a null statement, it does nothing when executed.


```

```python

# while loop: using break

a = 1
while a < 15:
    print(a)
    a = a + 4

    if a == 10:
        break
    
```

```python
# while loop: using continue
# here when iteration reaches step at continue, it ignore the output of that step(s) and continue the rest.

z = 0
while z < 100:
       z = z + 20
       if z == 40:
              continue
       print(z)

```

```python
# while loop: using pass (just use it as a filler for some funcatioanlity and later can add more modifications)

a = 0
while a < 10:
    a = a + 1
    if a == 5:
        pass
    print(a)


```
