```python
"""
Functions are basically the essence of any programming language that can be used to reuse your code more efficiently.

syntax: def < funcation_name_123() >


"""

def helloworld():
    print("hello world, how are you doing today?")

helloworld()
```

```python
# lets create a simple function to call a name with Hi

def greetings(name):
    print("Hi " + name + ", how are you doing today?")

# lets call the function greetings with a name.
greetings("prad")
```

```python
# lets create a function to add 2 numbers.
def letsadd(a,b):
    print(a + b )

letsadd(249,793)
```

```python
# lets make use of return in the function.
# return lets the value of the function to be assigned to a variable.

def returnadd(num1,num2):
    return(num1 + num2)

sum = returnadd(129,734)

print(sum)

# so, once python see the return keyword, it will not go beyond this & ignore
```
