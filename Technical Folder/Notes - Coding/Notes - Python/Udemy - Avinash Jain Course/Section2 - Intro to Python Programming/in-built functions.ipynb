#### python's  In-built functions
#### there are several in-built functions in python for various tasks


```python

# abs(): absolute function
# it converts any type of negative to positive number
abs(-7236)
```

```python
# bool () : Bool function
# if the number passed is Zero (0) and NONE, it will return FALSE
# if the number passed is non-zero, it will return TRUE


# bool(0) 
# bool(None)
# bool(828)
```

```python
# dir () : can be used to check what all function can be used for the string,numbers are passed in the function.
dir("hello world")
# dir(123)
```

```python
# help function

# lets say, we have a string and wanted to know what type of functions are there for 

word1 = ("hello world")

# now, let use the string here with help () as shown.
help(word1.upper)

# so, the function uppercase can be used as shown below
# syntax: Variable.upper()
word1.upper()
```

```python
# eval() function: runs the string in the function as python code, it is suitable for 1 line of code.

eval('print("hi")')
eval('5*3')
```

```python
# exec() is similar to eval(), but it is widely used for multi-line code.

w2 = 'print("hello world , 1111 222,333")'
exec(w2)
```

```python
# many conversion functions for string, integer,float
a = 101

str(a)

int(a)

float(a)
```
