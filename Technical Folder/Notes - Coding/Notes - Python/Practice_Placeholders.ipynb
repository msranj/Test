# PlaceHolders in python  
Source: https://www.udemy.com/course/pythonbootcamp/learn/lecture/10868130#notes  

- basically placeholders are taking the place of something  


```python
name = 'jake'
# suppose we want to print out a string that says "jake is 15 years old"
name + ' is 15 years old'

#other way to do it is to use an placeholder.
sent = "%s is 15 years old"
sent%name # placeholder

# in place of name, we can use any specific name, for example, "prad"
sent%("prad")

# lets use 2 names in the string, for example, "jake" and "prad"
sent = "%s %s is the president of the company"
sent%("barack","obama")

# lets use 1 string and 1 interger in the string, for example, "jake" and 15
sent = "%s is %d yeards old"
sent%("steve", 25)


```
