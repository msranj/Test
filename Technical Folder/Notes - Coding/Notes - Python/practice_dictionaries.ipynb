```python
## Dictionaries
- uses key, value for storing data
- use curly brackets
```

```python
# lets store students name and their ID
students = {"adam":12,"steve":89,"stephy":32,"linda":76 }
students

# find the value of key
students ["steve"]
```

```python
# update the value of key
students ["adam"] = 11 # earlier it was 12
students["adam"]

# delete the specific key and value pair
del students["adam"]

students

# length of dictionary
len(students)

# avoid to use same values as python will not be able to differentiate between them and it rerieves last key pair value
# for instance if there are "adam":11,"adam":12,"adam":13, it will only retrieve 13 as the value of key "adam"
students= {"adam":11,"adam":12,"adam":13}
students["adam"]




```
