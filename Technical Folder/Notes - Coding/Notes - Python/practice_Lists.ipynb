# Lists
- is a data structure 
- its a set of items
- comma seperated values


```python
# lets create a shopping list as an example
shoplist = ["apples","oranges","bananas","cheese","milk"]
shoplist

# to access the list we can use indexing
shoplist[0] # will give us the 1st item in the list
shoplist[2] # will give us the 2nd item in the list
```

```python
# lets append few more items to the list
shoplist.append("bread") # will add these items to the end of the list
shoplist
```

```python
# lets drop an item from the list  
shoplist.remove("bread") # will remove the specified item from the list
shoplist.remove("cheese") # will remove the specified item from the list
shoplist
```
