```python
# Operators : Add, subtract, division, module

# addition
Age1 = int(10)
Age2 = int(20)


AddValue        = Age1+Age2
DiffValue       = Age2-Age1
MultipleValue   = Age1*Age2
divValue        = Age2/Age1
ModuleValue     = Age2 % Age1
ModuleValue     = Age1 % Age2

print(f'the addition of {Age1} + {Age2}       : {AddValue}')
print(f'the difference of {Age2} - {Age1}     : {DiffValue}')
print(f'the multiplication of {Age1} * {Age2} : {MultipleValue}')
print(f'the division of {Age2} / {Age1}       : {divValue}')
print(f'the division of {Age2} % {Age1}       : {ModuleValue}')
print(f'the division of {Age1} % {Age2}       : {ModuleValue}')

```
